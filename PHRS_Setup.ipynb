```python

import matplotlib.pyplot as plt

# Count the sleep disorder categories, including missing values
disorder_counts = saved_df["Sleep Disorder"].value_counts(dropna=False)

# Give missing values a clear label for the chart
disorder_counts.index = [
    "Missing" if pd.isna(category) else category
    for category in disorder_counts.index
]

# Create a bar chart
disorder_counts.plot(kind="bar")

# Add chart title and labels
plt.title("Sleep Disorder Distribution")
plt.xlabel("Sleep Disorder Category")
plt.ylabel("Number of People")
plt.xticks(rotation=0)

# Display the chart
plt.show()
```

```python

import matplotlib.pyplot as plt

# Calculate average sleep duration for each BMI category
sleep_by_bmi = saved_df.groupby("BMI Category")["Sleep Duration"].mean()

# Create a bar chart
sleep_by_bmi.plot(kind="bar")

# Add chart title and labels
plt.title("Average Sleep Duration by BMI Category")
plt.xlabel("BMI Category")
plt.ylabel("Average Sleep Duration (hours)")
plt.xticks(rotation=0)

# Display the chart
plt.show()
```

```python

import matplotlib.pyplot as plt

# Calculate average stress level for each occupation
stress_by_occupation = saved_df.groupby("Occupation")["Stress Level"].mean()

# Sort from highest to lowest
stress_by_occupation = stress_by_occupation.sort_values(ascending=False)

# Create a bar chart
stress_by_occupation.plot(kind="bar", figsize=(10, 5))

# Add chart title and labels
plt.title("Average Stress Level by Occupation")
plt.xlabel("Occupation")
plt.ylabel("Average Stress Level")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()

# Display the chart
plt.show()
```

```python
import matplotlib.pyplot as plt

# Calculate average daily steps for each BMI category
steps_by_bmi = saved_df.groupby("BMI Category")["Daily Steps"].mean()

# Sort from highest to lowest
steps_by_bmi = steps_by_bmi.sort_values(ascending=False)

# Display the results
print(steps_by_bmi)

# Create a bar chart
steps_by_bmi.plot(kind="bar", figsize=(8, 5))

plt.title("Average Daily Steps by BMI Category")
plt.xlabel("BMI Category")
plt.ylabel("Average Daily Steps")
plt.xticks(rotation=0)
plt.tight_layout()

plt.show()
```

```python
import pandas as pd

file_path = r"C:\Users\User\OneDrive\Desktop\Personalized_Healthcare_System\data\healthcare_cleaned.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Number of records:", len(df))
print(df[["Person ID", "Sleep Duration"]].head())
```

```python
print("Minimum sleep duration:", df["Sleep Duration"].min())
print("Maximum sleep duration:", df["Sleep Duration"].max())
print("Missing sleep values:", df["Sleep Duration"].isna().sum())
```

```python
def sleep_duration_flag(hours):
    if hours < 7:
        return "Below Target"
    elif hours <= 9:
        return "Within Target"
    else:
        return "Above Target"

df["Sleep Duration Flag"] = df["Sleep Duration"].apply(sleep_duration_flag)

print(df[["Person ID", "Sleep Duration", "Sleep Duration Flag"]].head(10))
```

```python
sleep_flag_counts = df["Sleep Duration Flag"].value_counts()

print(sleep_flag_counts)
```
